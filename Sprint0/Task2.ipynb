Ниже в блоке реализуйте принцип Single Responsibility Principle (Принцип единственной обязанности) на базе класса Animal

```polyglot-notebook
public class Animal
{
    public virtual void Move()
    {
        Console.WriteLine("Животное двигается...");
    }
}

public interface ISoundable
{
    void MakeSound();
}

public class Dog : Animal, ISoundable
{
    public void MakeSound()
    {
        Console.WriteLine("Собака лает...");
    }
}

Dog myDog = new Dog();

myDog.Move();
myDog.MakeSound();
```

Задание:
Ниже в блоке реализуйте принцип Open/Closed Principle (Принцип открытости/закрытости) на базе класса Animal

```polyglot-notebook
public interface IMovable
{
    void Move();
}

public class Animal : IMovable
{
    public void Move()
    {
        Console.WriteLine("Животное двигается...");
    }
}

public class Dog : IMovable
{
    public void Move()
    {
        Console.WriteLine("Собака бегает...");
    }

    public void MakeSound()
    {
        Console.WriteLine("Собака лает...");
    }
}

Animal myAnimal = new Animal();
Dog myDog = new Dog();

myAnimal.Move();
myDog.Move();
myDog.MakeSound();
```

Задание:
Ниже в блоке реализуйте принцип Liskov Substitution Principle (Принцип подстановки Лисков) на базе класса Animal

```polyglot-notebook
public interface ILandAnimal
{
    void Move();
}

public interface IWaterAnimal
{
    void Swim();
}

public class Dog : ILandAnimal
{
    public void Move()
    {
        Console.WriteLine("Собака бегает...");
    }

    public void MakeSound()
    {
        Console.WriteLine("Собака лает...");
    }
}

public class Fish : IWaterAnimal
{
    public void Swim()
    {
        Console.WriteLine("Рыба плавает...");
    }
}

Dog myDog = new Dog();
Fish myFish = new Fish();

myDog.Move();
myFish.Swim();
```

Задание:
Ниже в блоке реализуйте принцип Interface Segregation Principle (Принцип разделения интерфейсов) на базе класса Animal

```polyglot-notebook
public interface IAnimal
{
    void Move();
}

public interface ICanMakeSound
{
    void MakeSound();
}

public interface ICanSwim
{
    void Swim();
}

public interface ICanFly
{
    void Fly();
}

public class Dog : IAnimal, ICanMakeSound
{
    public void Move()
    {
        Console.WriteLine("Собака двигается...");
    }

    public void MakeSound()
    {
        Console.WriteLine("Собака лает...");
    }
}

public class Fish : IAnimal, ICanSwim
{
    public void Move()
    {
        Console.WriteLine("Рыба двигается...");
    }

    public void Swim()
    {
        Console.WriteLine("Рыба плавает...");
    }
}

Dog myDog = new Dog();
Fish myFish = new Fish();

myDog.Move();
myDog.MakeSound();

myFish.Move();
myFish.Swim();
```

Задание:
Ниже в блоке реализуйте принцип Dependency Inversion Principle (Принцип инверсии зависимостей) на базе класса Animal

```polyglot-notebook
public interface IFood
{
    void Eat();
}

public class Meat : IFood
{
    public void Eat()
    {
        Console.WriteLine("Животное ест мясо...");
    }
}

public class Animal
{
    private IFood _food;

    public Animal(IFood food)
    {
        _food = food;
    }

    public void EatFood()
    {
        _food.Eat();
    }
}

IFood food = new Meat();
Animal myAnimal = new Animal(food);

myAnimal.EatFood();
```
