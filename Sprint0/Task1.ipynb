2. Инкапсуляция Задание:

```polyglot-notebook
// Создание класса
class Animal
{
    private string name; // Приватное поле

    public string Name // Свойство с геттером и сеттером
    {
        get { return name; }
        set { name = value; }
    }

    public void Eat() // Публичный метод
    {
        Console.WriteLine("Животное кушает...");
    }
}

// Основная программа
Animal MyAnimal = new Animal(); // Создание объекта
MyAnimal.Name = "Анфиса";
MyAnimal.Eat();
Console.WriteLine(MyAnimal.Name);
```

3. Наследование Задание:
Ниже в блоке по примеру создайте класс Animal и реализуйте наследование

```polyglot-notebook
class Animal
{
    private string name;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public void Move()
    {
        Console.WriteLine("Животное двигается...");
    }
}

class Dog : Animal
{
}

Animal myAnimal = new Animal();
Dog myDog = new Dog();

myDog.Name = "Бобик";
Console.WriteLine(myDog.Name);
myDog.Move();

myAnimal.Name = "Животное";
Console.WriteLine(myAnimal.Name);
myAnimal.Move();
```

```polyglot-notebook
class Animal
{
    public virtual void MakeSound()
    {
        Console.WriteLine("Животное издает звук");
    }
}

class Dog : Animal
{
    public override void MakeSound()
    {
        Console.WriteLine("Собака лает");
    }
}

Animal myAnimal = new Animal();
Dog myDog = new Dog();

myAnimal.MakeSound();
myDog.MakeSound();
```

5. Абстракция адание:
Ниже в блоке по примеру создайте класс Animal и реализуйте абстракцию

```polyglot-notebook
abstract class Animal
{
    public abstract void MakeSound();

    public void Move()
    {
        Console.WriteLine("Животное двигается...");
    }
}

class Dog : Animal
{
    public override void MakeSound()
    {
        Console.WriteLine("Собака лает...");
    }
}

Dog myDog = new Dog();

myDog.MakeSound();
myDog.Move();
```

6. Ассоциация (Композиция и Агрегация) Задание:
Ниже в блоке по примеру создайте класс Animal и реализуйте композицию

```polyglot-notebook
class Heart
{
    public void Beat()
    {
        Console.WriteLine("Сердце бьется");
    }
}

class Animal
{
    private Heart heart;

    public Animal(Heart heart)
    {
        this.heart = heart;
    }

    public void Live()
    {
        heart.Beat();
        Console.WriteLine("Животное живет");
    }
}

Heart myHeart = new Heart();
Animal myAnimal = new Animal(myHeart);

myAnimal.Live();
```

7. Ассоциация (Композиция и Агрегация) Задание:
Ниже в блоке по примеру создайте класс Animal и реализуйте агрегацию

```polyglot-notebook
abstract class LivingBeing
{
    public abstract void Live();
}

class Animal : LivingBeing
{
    private Owner owner;

    public Animal(Owner owner)
    {
        this.owner = owner;
    }

    public override void Live()
    {
        Console.WriteLine("Животное живет...");
    }
}

class Owner
{
    public string Name { get; set; }

    public Owner(string name)
    {
        Name = name;
    }

    public void Care(LivingBeing livingBeing)
    {
        Console.WriteLine($"{Name} заботится о животном...");
        livingBeing.Live();
    }
}

Owner maxim = new Owner("Максим");
Animal myAnimal = new Animal(maxim);

myAnimal.Live();
maxim.Care(myAnimal);
```
