<h2 style="color:DodgerBlue">Реализация:</h2>

```polyglot-notebook
// 1. Создание объектов
Conference conf = new Conference(1, "IT-Конференция", new DateTime(2024, 5, 20), 5);
Festival fest = new Festival(2, "Рок-Фестиваль", new DateTime(2024, 7, 15), 10);
Seminar sem = new Seminar(3, "Семинар по C#", new DateTime(2024, 6, 10), new List<string> { "ООП", "LINQ", "Асинхронность" });

// Создаем объект сложного наследования (HybridEvent -> Conference -> Event)
HybridEvent hybrid = new HybridEvent(4, "Гибридная конференция", new DateTime(2024, 8, 1), 3, "https://live.stream.com", "TechCorp");

// 2. Вывод деталей мероприятий
Console.WriteLine("--- Информация о мероприятиях ---");
conf.GetEventDetails();
Console.WriteLine($"Количество спикеров: {conf.SpeakersCount}\n");

fest.GetEventDetails();
Console.WriteLine($"Количество групп: {fest.BandsCount}\n");

sem.GetEventDetails(); // Здесь сработает переопределенный метод
Console.WriteLine();

hybrid.GetEventDetails();
Console.WriteLine($"URL трансляции: {hybrid.StreamUrl}, Спонсор: {hybrid.SponsorName}\n");

// 3. Проверка множественного наследования (методы интерфейсов)
Console.WriteLine("--- Проверка интерфейсов (множественное наследование) ---");
hybrid.StartStream();
hybrid.AddSponsor("Global Sponsor");

// 4. Демонстрация полиморфизма (уведомления и регистрация)
Console.WriteLine("\n--- Демонстрация полиморфизма ---");
conf.NotifyParticipants(); // Уведомление с программой
fest.NotifyParticipants(); // Базовое уведомление

conf.RegisterParticipant(); // Базовая регистрация
fest.RegisterParticipant(); // Регистрация с выбором жанра


// =========================================================================
// ИНТЕРФЕЙСЫ (ДЛЯ МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ)
// =========================================================================

public interface IStreamable
{
    void StartStream();
}

public interface ISponsorable
{
    void AddSponsor(string sponsorName);
}


// =========================================================================
// БАЗОВЫЙ КЛАСС
// =========================================================================

public class Event
{
    private int _eventId;
    private string _eventName;
    private DateTime _date;
    private decimal _baseTicketPrice; // Добавлен атрибут 1: базовая цена билета

    public int EventId
    {
        get { return _eventId; }
        set { _eventId = value; }
    }

    public string EventName
    {
        get { return _eventName; }
        set { _eventName = value; }
    }

    public DateTime Date
    {
        get { return _date; }
        set { _date = value; }
    }

    public decimal BaseTicketPrice
    {
        get { return _baseTicketPrice; }
        set { _baseTicketPrice = value; }
    }

    // Конструктор базового класса
    public Event(int eventId, string eventName, DateTime date, decimal baseTicketPrice)
    {
        EventId = eventId;
        EventName = eventName;
        Date = date;
        BaseTicketPrice = baseTicketPrice;
    }

    public virtual void RegisterParticipant()
    {
        Console.WriteLine($"Участник зарегистрирован на мероприятие '{EventName}'.");
    }

    public virtual void NotifyParticipants()
    {
        Console.WriteLine($"Уведомление: Мероприятие '{EventName}' состоится {Date.ToShortDateString()}.");
    }

    public virtual void GetEventDetails()
    {
        Console.WriteLine($"ID: {EventId} | Название: {EventName} | Дата: {Date.ToShortDateString()} | Цена билета: {BaseTicketPrice} руб.");
    }

    // Добавлен метод 1: расчет стоимости
    public decimal CalculateTotalCost(int participantsCount)
    {
        return participantsCount * BaseTicketPrice;
    }
}


// =========================================================================
// ПРОСТОЕ НАСЛЕДОВАНИЕ (ПРОИЗВОДНЫЕ КЛАССЫ)
// =========================================================================

// 1. Конференция (+ множественное наследование через IStreamable)
public class Conference : Event, IStreamable
{
    private int _speakersCount;

    public int SpeakersCount
    {
        get { return _speakersCount; }
        set { _speakersCount = value; }
    }

    public Conference(int eventId, string eventName, DateTime date, int speakersCount)
        : base(eventId, eventName, date, 1000) // Фиксированная цена для конференции
    {
        SpeakersCount = speakersCount;
    }

    // Переопределение метода NotifyParticipants
    public override void NotifyParticipants()
    {
        Console.WriteLine($"Уведомление для конференции '{EventName}': Ознакомьтесь с программой конференции. Спикеров: {SpeakersCount}.");
    }

    // Метод интерфейса IStreamable
    public void StartStream()
    {
        Console.WriteLine($"Трансляция конференции '{EventName}' началась.");
    }
}

// 2. Фестиваль
public class Festival : Event
{
    private int _bandsCount;

    public int BandsCount
    {
        get { return _bandsCount; }
        set { _bandsCount = value; }
    }

    public Festival(int eventId, string eventName, DateTime date, int bandsCount)
        : base(eventId, eventName, date, 500)
    {
        BandsCount = bandsCount;
    }

    // Переопределение метода RegisterParticipant
    public override void RegisterParticipant()
    {
        Console.WriteLine($"Регистрация на фестиваль '{EventName}': Выберите интересующие музыкальные жанры (Рок, Поп, Джаз).");
    }
}

// 3. Семинар
public class Seminar : Event
{
    private List<string> _topics;

    public List<string> Topics
    {
        get { return _topics; }
        set { _topics = value; }
    }

    public Seminar(int eventId, string eventName, DateTime date, List<string> topics)
        : base(eventId, eventName, date, 300)
    {
        Topics = topics;
    }

    // Переопределение метода GetEventDetails
    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Тематические области: {string.Join(", ", Topics)}");
    }
}


// =========================================================================
// СЛОЖНОЕ НАСЛЕДОВАНИЕ (HybridEvent -> Conference -> Event)
// + МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ (реализует еще и ISponsorable)
// =========================================================================

public class HybridEvent : Conference, ISponsorable
{
    private string _streamUrl;
    private string _sponsorName;

    public string StreamUrl
    {
        get { return _streamUrl; }
        set { _streamUrl = value; }
    }
    
    public string SponsorName
    {
        get { return _sponsorName; }
        set { _sponsorName = value; }
    }

    public HybridEvent(int eventId, string eventName, DateTime date, int speakersCount, string streamUrl, string sponsorName)
        : base(eventId, eventName, date, speakersCount)
    {
        StreamUrl = streamUrl;
        SponsorName = sponsorName;
    }

    // Метод интерфейса ISponsorable
    public void AddSponsor(string sponsorName)
    {
        SponsorName = sponsorName;
        Console.WriteLine($"Спонсор '{sponsorName}' добавлен к мероприятию '{EventName}'.");
    }
}


// =========================================================================
// КЛАСС ДЛЯ ВЗАИМОДЕЙСТВИЯ (Аналог Order)
// =========================================================================

public class EventManager
{
    public void ManageEvent(Event ev)
    {
        Console.WriteLine($"\nУправление мероприятием: {ev.EventName}");
        ev.NotifyParticipants();
        ev.RegisterParticipant();
    }
}
```
